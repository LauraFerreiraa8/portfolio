```python
%%writefile DesafioBusca.java
import java.util.Arrays;
import java.util.Random;
import java.util.Scanner;

public class DesafioBusca {

    public static void main(String[] args) {
        Scanner leitor = new Scanner(System.in);
        int tamanhoLista = 1000000; // 1 Milhão de elementos
        int[] lista = new int[tamanhoLista];
        Random gerador = new Random();

        // 1. Preparando a lista (Ordenada para que a Binária funcione)
        for (int i = 0; i < tamanhoLista; i++) {
            lista[i] = i * 2; // Gera apenas números pares: 0, 2, 4, 6...
        }

        System.out.println("=== DESAFIO DE ALGORITMOS EM JAVA ===");
        System.out.println("A lista tem " + tamanhoLista + " números pares (de 0 a 1.999.998).");
        System.out.print("Digite um número para buscar: ");
        int alvo = leitor.nextInt();

        System.out.println("\n--- RESULTADOS ---");

        // Executando Busca Linear
        int passosLinear = buscaLinear(lista, alvo);
        System.out.println("Busca Linear: " + (passosLinear != -1 ?
            "Encontrado em " + passosLinear + " passos." : "Não encontrado."));

        // Executando Busca Binária
        int passosBinaria = buscaBinaria(lista, alvo);
        System.out.println("Busca Binária: " + (passosBinaria != -1 ?
            "Encontrado em " + passosBinaria + " passos." : "Não encontrado."));

        leitor.close();
    }

    /**
     * Busca Linear: Olha um por um.
     * Retorna o número de passos (iterações) realizados.
     */
    public static int buscaLinear(int[] lista, int alvo) {
        int passos = 0;
        for (int i = 0; i < lista.length; i++) {
            passos++;
            if (lista[i] == alvo) {
                return passos;
            }
        }
        return -1; // Não encontrado
    }

    /**
     * Busca Binária: Divide a lista ao meio.
     * Retorna o número de passos (divisões) realizadas.
     */
    public static int buscaBinaria(int[] lista, int alvo) {
        int passos = 0;
        int inicio = 0;
        int fim = lista.length - 1;

        while (inicio <= fim) {
            passos++;
            int meio = inicio + (fim - inicio) / 2;

            if (lista[meio] == alvo) {
                return passos;
            }

            if (lista[meio] < alvo) {
                inicio = meio + 1;
            } else {
                fim = meio - 1;
            }
          System.out.println("Passos:  " + passos);
        }
        return -1; // Não encontrado
    }
}
```

```python
%%writefile AtividadeBuscaLinear.java
import java.util.Scanner;
import java.util.Random;

public class AtividadeBuscaLinear {

    public static void main(String[] args) {
        Scanner leitor = new Scanner(System.in);
        Random gerador = new Random();

        // Criando um array de 10 posições
        int[] numeros = new int[10];

        // Preenchendo com números aleatórios de 1 a 20
        System.out.println("--- LISTA GERADA ---");
        for (int i = 0; i < numeros.length; i++) {
            numeros[i] = gerador.nextInt(20) + 1;
            System.out.print("[" + numeros[i] + "] ");
        }

        System.out.print("\n\nQual número você deseja buscar? ");
        int alvo = leitor.nextInt();

        // Chamada do método de busca
        int posicaoEncontrada = realizarBuscaLinear(numeros, alvo);

        // Exibição do Resultado
        if (posicaoEncontrada != -1) {
            System.out.println("✅ Sucesso! O número " + alvo + " foi encontrado na posição (índice): " + posicaoEncontrada);
        } else {
            System.out.println("❌ Ops! O número " + alvo + " não está na lista.");
        }

        leitor.close();
    }

    /**
     * Método que percorre o array do início ao fim.
     * @return O índice onde o número está, ou -1 se não encontrar.
     */
    public static int realizarBuscaLinear(int[] lista, int numeroProcurado) {
        for (int i = 0; i < lista.length; i++) {
            // Mostrando o "pensamento" do computador
            System.out.println("Verificando índice " + i + ": O valor é " + lista[i] + "?");

            if (lista[i] == numeroProcurado) {
                return i; // Retorna o índice assim que encontra
            }
        }
        return -1; // Se percorrer tudo e não achar, retorna -1
    }
}
```

```python
import java.util.Arrays;
import java.util.Scanner;

public class AtividadeBuscaBinaria {

    public static void main(String[] args) {
        Scanner leitor = new Scanner(System.in);

        // 1. Criando um array de 15 elementos
        int[] lista = {2, 5, 8, 12, 16, 23, 38, 45, 56, 67, 78, 81, 91, 95, 99};

        System.out.println("--- LISTA ORDENADA ---");
        System.out.println(Arrays.toString(lista));

        System.out.print("\nDigite o número que deseja encontrar: ");
        int alvo = leitor.nextInt();

        // 2. Chamada do método
        int resultado = realizarBuscaBinaria(lista, alvo);

        // 3. Resultado final
        if (resultado != -1) {
            System.out.println("\n✅ Encontrado! O número " + alvo + " está no índice " + resultado + ".");
        } else {
            System.out.println("\n❌ O número " + alvo + " não existe nesta lista.");
        }

        leitor.close();
    }

    public static int realizarBuscaBinaria(int[] array, int alvo) {
        int inicio = 0;
        int fim = array.length - 1;
        int tentativas = 0;

        while (inicio <= fim) {
            tentativas++;
            int meio = inicio + (fim - inicio) / 2;

            System.out.println("Tentativa " + tentativas + ": Verificando o meio (índice " + meio + ", valor " + array[meio] + ")");

            // Caso 1: Encontrou de primeira!
            if (array[meio] == alvo) {
                return meio;
            }

            // Caso 2: O alvo é MAIOR que o meio? Descarta a metade esquerda
            if (array[meio] < alvo) {
                System.out.println("   -> " + alvo + " é maior que " + array[meio] + ". Indo para a DIREITA.");
                inicio = meio + 1;
            }
            // Caso 3: O alvo é MENOR que o meio? Descarta a metade direita
            else {
                System.out.println("   -> " + alvo + " é menor que " + array[meio] + ". Indo para a ESQUERDA.");
                fim = meio - 1;
            }
        }

        return -1; // Se o loop acabar e não encontrar
    }
}
```
